# Linear Regression on Mushroom Dataset (Demo)
Theory: Linear regression predicts a continuous target as a linear combination of input features. The mushroom dataset target `class` is categorical (edible/poisonous); for a demo we map it to numeric (0/1) and treat it as a continuous target — this is a pedagogical example only. For classification use LogisticRegression or Naive Bayes.
Syntax: We'll use `sklearn.linear_model.LinearRegression`, `mean_squared_error` and `r2_score` for evaluation.

# Imports
Theory: Import pandas for data handling and scikit-learn for modeling and metrics.
Syntax: `from sklearn.linear_model import LinearRegression`

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

# Load dataset
Theory: Load the mushroom CSV into a DataFrame and inspect basic properties.
Syntax: `pd.read_csv(path, encoding='unicode_escape')`

```python
df = pd.read_csv('../mushroom/mushrooms.csv', encoding='unicode_escape')
display(df.head())
print('shape:', df.shape)
print('\nclass distribution:\n', df['class'].value_counts())
```

# Preprocessing
Theory: One-hot encode categorical features so they can be used by a linear model.
Syntax: `pd.get_dummies(df.drop('class', axis=1))`

```python
X = pd.get_dummies(df.drop('class', axis=1), drop_first=False)
X.shape
```

# Target
Theory: Map `class` to numeric values (e=0, p=1) and use as the regression target for this demo.
Syntax: `y = df['class'].map({'e':0,'p':1})`

```python
y = df['class'].map({'e': 0, 'p': 1})
y.value_counts()
```

# Train/test split
Theory: Split data to evaluate generalization. Syntax: `train_test_split(X,y,test_size=0.2,random_state=11)`

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=11)
print('train shape:', X_train.shape, 'test shape:', X_test.shape)
```

# Model
Theory: Fit a LinearRegression model (ordinary least squares).
Syntax: `LinearRegression()`

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
```

# Predictions
Predict continuous outputs on the test set. Syntax: `y_pred = lr.predict(X_test)`

```python
y_pred = lr.predict(X_test)
# show a few predictions vs actual
pd.DataFrame({'y_true': y_test, 'y_pred': y_pred}).head()
```

# Evaluation
Theory: Use mean squared error (MSE) and R^2 to evaluate regression performance. For this demo target R^2 will illustrate how well continuous approximation performs on a categorical target.
Syntax: `mean_squared_error(y_true, y_pred)` and `r2_score(y_true, y_pred)`

```python
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print('MSE:', mse)
print('R^2:', r2)
```
